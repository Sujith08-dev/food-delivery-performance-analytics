```python
import pandas as pd

# 1. Reload the ORIGINAL train.csv
df = pd.read_csv('/content/train.csv')

# 2. Remove extra spaces from column names
df.columns = df.columns.str.strip()

# 3. Clean whitespace from all text columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# 4. Convert fake missing values to real NaN
df = df.replace(['NaN', 'nan', 'NaN '], pd.NA)

# 5. Numeric columns
numeric_cols = [
    'Delivery_person_Age',
    'Delivery_person_Ratings',
    'multiple_deliveries'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 6. FIX Time_taken(min)
# Values are like "(min) 24", so remove "(min)"
df['Time_taken(min)'] = (
    df['Time_taken(min)']
    .astype('string')
    .str.replace('(min)', '', regex=False)
    .str.strip()
)

df['Time_taken(min)'] = pd.to_numeric(
    df['Time_taken(min)'],
    errors='coerce'
)

# 7. Convert date
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    errors='coerce'
)

# 8. Convert order times
for col in ['Time_Orderd', 'Time_Order_picked']:
    df[col] = pd.to_datetime(
        df[col],
        format='%H:%M:%S',
        errors='coerce'
    ).dt.time

# 9. Ratings should be between 1 and 5
df.loc[
    (df['Delivery_person_Ratings'] < 1) |
    (df['Delivery_person_Ratings'] > 5),
    'Delivery_person_Ratings'
] = pd.NA

# 10. Check duplicates
df = df.drop_duplicates()

# 11. Final validation
print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nTime Taken sample:")
print(df['Time_taken(min)'].head(10))

print("\nRating values:")
print(sorted(df['Delivery_person_Ratings'].dropna().unique()))
```

```python
import pandas as pd

# ==========================================
# 1. LOAD ORIGINAL DATA
# ==========================================
df = pd.read_csv('/content/train.csv')

# Clean column names
df.columns = df.columns.str.strip()

# Remove leading/trailing spaces from text columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# Convert common fake missing values to actual NaN
df = df.replace(['NaN', 'nan', ''], pd.NA)


# ==========================================
# 2. NUMERIC COLUMNS
# ==========================================
numeric_cols = [
    'Delivery_person_Age',
    'Delivery_person_Ratings',
    'multiple_deliveries'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')


# ==========================================
# 3. DELIVERY TIME
# ==========================================
# Original values look like "(min) 24"
df['Time_taken(min)'] = (
    df['Time_taken(min)']
    .astype('string')
    .str.replace('(min)', '', regex=False)
    .str.strip()
)

df['Time_taken(min)'] = pd.to_numeric(
    df['Time_taken(min)'],
    errors='coerce'
)


# ==========================================
# 4. DATE
# ==========================================
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    format='%d-%m-%Y',
    errors='coerce'
)


# ==========================================
# 5. TIME COLUMNS
# ==========================================
for col in ['Time_Orderd', 'Time_Order_picked']:
    df[col] = pd.to_datetime(
        df[col],
        format='%H:%M:%S',
        errors='coerce'
    ).dt.time


# ==========================================
# 6. INVALID RATINGS
# ==========================================
# Valid rating range = 1 to 5
df.loc[
    (df['Delivery_person_Ratings'] < 1) |
    (df['Delivery_person_Ratings'] > 5),
    'Delivery_person_Ratings'
] = pd.NA


# ==========================================
# 7. HANDLE MISSING CATEGORICAL VALUES
# ==========================================
categorical_cols = [
    'Weatherconditions',
    'Road_traffic_density',
    'Festival',
    'City'
]

for col in categorical_cols:
    df[col] = df[col].fillna('Unknown')


# ==========================================
# 8. HANDLE MISSING NUMERIC VALUES
# ==========================================
# Median is less affected by extreme values than mean
for col in [
    'Delivery_person_Age',
    'Delivery_person_Ratings',
    'multiple_deliveries'
]:
    df[col] = df[col].fillna(df[col].median())


# ==========================================
# 9. MISSING ORDER TIME
# ==========================================
df['Time_Orderd'] = df['Time_Orderd'].fillna(
    df['Time_Order_picked']
)


# ==========================================
# 10. REMOVE DUPLICATES
# ==========================================
df = df.drop_duplicates()


# ==========================================
# 11. FINAL VALIDATION
# ==========================================
print("========== FINAL DATASET ==========")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

print("\n========== DUPLICATES ==========")
print(df.duplicated().sum())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== RATING RANGE ==========")
print(df['Delivery_person_Ratings'].min(),
      "to",
      df['Delivery_person_Ratings'].max())

print("\n========== DELIVERY TIME ==========")
print(df['Time_taken(min)'].min(),
      "to",
      df['Time_taken(min)'].max())


# ==========================================
# 12. EXPORT CLEAN DATA
# ==========================================
df.to_csv('/content/cleaned_train.csv', index=False)

print("\n✅ cleaned_train.csv created successfully!")
```
